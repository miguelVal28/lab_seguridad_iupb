## 3. Injection

### Descripción

Las inyecciones ocurren cuando un atacante puede enviar código malicioso a través de una entrada de usuario para ser ejecutado por la aplicación.

### Ejemplo de Vulnerabilidad

Un ejemplo clásico es la inyección SQL, donde un atacante puede manipular una consulta SQL.

```python
# Entrada de usuario insegura
user_input = "1; DROP TABLE users;"

# Consulta vulnerable
query = f"SELECT * FROM users WHERE id = {user_input}"
```

### Simulación de Auditoría Automatizada con SQLMap

**sqlmap** es una herramienta de código abierto que automatiza el proceso de detectar y explotar fallos de inyección SQL en aplicaciones web. En este ejercicio, simularemos cómo un atacante o un auditor de seguridad utiliza esta herramienta desde la consola de comandos dentro de la infraestructura en la nube de Google Colab.

#### Argumentos clave utilizados en el script:
* **`-u`**: Especifica la URL del objetivo (Target).
* **`--batch`**: Ejecuta la herramienta en modo automático, respondiendo "sí" por defecto a todas las preguntas del sistema.
* **`--level=1` y `--risk=1`**: Ajustan la profundidad y el riesgo de las pruebas a un entorno controlado para optimizar el tiempo de ejecución en la sesión.
